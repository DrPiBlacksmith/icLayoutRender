The `icLayoutRender` notebook will render the layout for the cell specified in `cell`. In this example, the target cell is named **Neuron**.

---

## GDS to PDF Conversion Example

This Jupyter Notebook demonstrates how to convert a GDSII cell into a PDF using the `gds2pdf` parser from the [icLayoutRender](https://github.com/DrPiBlacksmith/icLayoutRender) project.  

The example uses **STB55 technology layers** and includes **layer color mapping**. The target cell, **Neuron**, is available in the `./Examples/` folder.

---

## Setup

1. Place your GDSII file in the `./Examples/` folder.  
2. Ensure you have a layer color map file for STB55 layers, e.g., `LayerColors_STB55.map`.  
3. Activate LaTeX PDF compilation if desired by setting `pdfTex = True`.  

---

## Notes

- Make sure the GDS file exists in the `./Examples/` folder.  
- Layer colors must match the technology layers for proper visualization.  
- Ensure that the color map matches your technology. You may need to create a custom map to perfectly match your desired colors.  
- Have fun exploring the PDF output, or convert it to **SVG format** for a more versatile layout representation.  

```python
# Library definitions 
import gdspy
import pandas as pd
import math
from gds2pdf import gds2pdf
from GDSLatexConverter import GDSLatexConverter
import os
import sys, getopt
from IPython.display import display, IFrame

cell = "Neuron" #This cell is generated in STB55 tech layers
cellName = print("./Examples/%s.gds",cell)
pdfName = print("./Examples/%s.pdf",cell)
layerColors = "./LayerColors_STB55.map" #Correct layer color map for STB55
pdfTex = True #pdftex compiler option is activated


gds2pdf(cellName,pdfName,layerColors,pdfTex)

# Display it inline the PDF output (width and height are adjustable)
display(IFrame(pdfName, width=400, height=300))
```
